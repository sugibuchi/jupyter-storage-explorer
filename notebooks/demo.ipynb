# Storage Explorer demo

Browses `samples/` (a small data lake made by `make_samples.py`) and an in-memory scratch space; switch with the drop-down left of the address bar.

- **Browse:** click a folder in the tree; double-click a folder to open it. Ctrl+click selects several items.
- **Preview:** double-click a file, or use **Preview** (with nothing selected, it previews the current folder, e.g. a Delta table).
- **Right-click:** Preview as, Edit (text files), Python code to load the data, Download, Copy path / URL.
- **Drag and drop:** drop files from JupyterLab's file browser or your computer to upload; drag files onto a folder to move them.

Edits, uploads and moves change `samples/`; run `make_samples.py` to restore it.

```python
import fsspec

from jupyter_storage_explorer import StorageExplorer

# A second filesystem to switch to with the drop-down left of the address bar:
# an in-memory scratch space with a few files.
scratch = fsspec.filesystem("memory")
scratch.pipe(
    {
        "/scratch/hello.txt": b"Hello from memory://\n",
        "/scratch/squares.csv": b"n,square\n1,1\n2,4\n3,9\n",
    }
)

# height: the tree and file list; the preview panel adds its own height below.
explorer = StorageExplorer(
    filesystems={"Samples (local)": "samples", "Scratch (memory)": scratch},
    height="350px",
)
explorer
```
