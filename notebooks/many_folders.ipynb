# Many folders

An in-memory filesystem with 200 folders, to check that the tree on the left and the file list scroll:

- **Vertically:** the 200 folders overflow both the tree and the list (`height="300px"`).
- **Sideways:** `folder-001` holds 12 nested levels, and `folder-002` a folder with a very long name; expand them in the tree.
- **Resizing:** drag the tree's bottom-right corner to make it wider, and the bar under the list to make both taller.

```python
import fsspec

from jupyter_storage_explorer import StorageExplorer

fs = fsspec.filesystem("memory")
# The memory filesystem is shared by the whole kernel: start afresh on every run.
if fs.exists("/many"):
    fs.rm("/many", recursive=True)

# 200 folders, each with a small file.
fs.pipe({f"/many/folder-{i:03d}/data.csv": b"a,b\n1,2\n" for i in range(1, 201)})
# Deep and long-named folders, for sideways scrolling in the tree.
deep = "/".join(f"level-{d:02d}" for d in range(1, 13))
fs.pipe(
    {
        f"/many/folder-001/{deep}/deep.txt": b"deep\n",
        "/many/folder-002/a-folder-with-a-very-long-name-that-does-not-fit-in-the-tree/x.txt": b"x\n",
    }
)

explorer = StorageExplorer("/many", fs=fs, height="300px")
explorer
```
